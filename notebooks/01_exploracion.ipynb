```python
import re
import unicodedata

def limpiar_nombre(c):
    c = c.replace("¥", "ñ")
    c = unicodedata.normalize("NFKD", c).encode("ascii", "ignore").decode()
    c = c.lower().strip()
    return re.sub(r"[^a-z0-9]+", "_", c).strip("_")

df.columns = [limpiar_nombre(c) for c in df.columns]
df = df.rename(columns={"tipo_escenenario": "tipo_escenario"})
df.columns.tolist()
```

```python
import numpy as np

df_clean = df.copy()

# Quitar columnas sin utilidad
df_clean = df_clean.drop(columns=["id", "futsala"])

# Limpiar espacios en todas las columnas de texto
cols_texto = df_clean.select_dtypes(exclude="number").columns
for c in cols_texto:
    df_clean[c] = df_clean[c].str.strip().str.replace(r"\s+", " ", regex=True)

# Estandarizar mayúsculas y rellenar nulos
df_clean["tipo_escenario"] = df_clean["tipo_escenario"].str.upper()
df_clean["tipo_espacio"] = df_clean["tipo_espacio"].fillna("Sin dato")

# Zona: comunas 1-22 = urbana, 51-65 = rural (corregimientos)
df_clean["zona"] = np.where(df_clean["comuna"] <= 22, "Urbana", "Rural")

# Variables derivadas
cols_inst = ["grama_natural", "sintetica", "multiple", "baloncesto", "voleibol",
             "piscina_adultos", "piscina_ninos", "juegos_infantiles", "biosaludables",
             "pista_patinaje", "pista_de_trote", "pista_bmx", "tenis", "tejo",
             "hockey", "microfutb", "voley_playa"]

df_clean["total_instalaciones"] = df_clean[cols_inst].sum(axis=1)
df_clean["tipos_instalacion"] = (df_clean[cols_inst] > 0).sum(axis=1)

df_clean.to_csv("../data/clean/escenarios_deportivos_limpio.csv",
                index=False, encoding="utf-8")

print(df_clean.shape)
print(df_clean["zona"].value_counts())
df_clean.head()
```
